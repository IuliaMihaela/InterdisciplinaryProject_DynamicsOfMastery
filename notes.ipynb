```python
# check player trajectories for each experiments (to check the trajectory length)

import pandas as pd
from pathlib import Path

processed_dir = Path("data/processed")

for traj_file in processed_dir.rglob("player_trajectories.csv"):
    experiment_name = traj_file.parent.name
    df = pd.read_csv(traj_file)
    
    # Count the number of observations (rows) per player
    obs_counts = df.groupby("person_id").size()
    
    min_obs = obs_counts.min()
    max_obs = obs_counts.max()
    median_obs = obs_counts.median()
    under_100 = (obs_counts < 100).sum()
    
    print(f"Experiment: {experiment_name}")
    print(f"  Total Players: {len(obs_counts)}")
    print(f"  Min Solves:    {min_obs}")
    print(f"  Max Solves:    {max_obs}")
    print(f"  Median Solves: {median_obs}")
    print(f"  Players < 100: {under_100}")
    print("-" * 40)
```
